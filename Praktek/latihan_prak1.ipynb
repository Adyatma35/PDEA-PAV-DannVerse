```python
import pandas as pd
```

```python
df = pd.read_csv("latihan 1.csv")
display(df)
```

```python
df["Skor_Kehadiran"] = df["Skor_Kehadiran"].str.replace("%","").astype(float) / 100

df
```

```python
df["Total_Pembayaran"] = df["Total_Pembayaran"].str.replace(".","").astype(float)

df
```

```python
df["Jenis_Kelamin"].unique()
```

```python
df["Jenis_Kelamin"] = df["Jenis_Kelamin"].replace({
    "Laki-laki":"Mas",
    "Perempuan":"Mba",
    "L":"Mas", 
    "P":"Mba", 
    "Male":"Mas", 
    "Female":"Mba"})
df
```

```python
df["Status_Keaktifan"].unique()
```

```python
df["Status_Keaktifan"] = df["Status_Keaktifan"].str.lower()

df
```

```python
df["Status_Keaktifan"].unique()
```

```python
df["Nama_Lengkap"] = df["Nama_Lengkap"].str.capitalize().str.title()
df
```

```python
df["Kota_Asal"].unique()
```

```python
df["Kota_Asal"] = df["Kota_Asal"].str.capitalize().str.replace({
    "Ska" : "Surakarta",
    "Solo" : "Surakarta"
})
df
```

```python
nomor_telepon = df["Nomor_Telepon"].str.replace({
    "+62": "0",
    "abc": "",
    "INVALID_PHONE": ""
})
```
