### Fungsi Download Data

#### `yf.download(tickers, start=None, end=None, period="max", interval="1d", group_by="column")`

1. `start`/`end`: format 'YYYY-MM-DD'

2. `period`: Alternatif rentang waktu (1d, 5d, 1mo, 3mo, 6mo, 1y, 2y, 5y, ytd, max)

3. `interval`: Rentang waktu antar data (1m, 2m, 5m, 15m, 30m, 60m, 90m, 1h, 4h, 1d, 5d, 1wk, 1mo, 3mo)

4. `group_by`: Pengelompokan (column, ticker)

```python
Data = yf.download("GOTO.JK", period="3mo", interval="1d", group_by="ticker")

Data
```

```python
Data = yf.download(["GOTO.JK","GRAB","UBER"], period="3mo", interval="1d", group_by="ticker")

Data
```

```python
Data = yf.download(["GOTO.JK","GRAB","UBER"], start="2026-08-10", end="2026-11-11", interval="1d", group_by="ticker")

Data
```
