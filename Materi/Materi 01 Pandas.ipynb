```python
data_json2 = pd.read_json("./Dataset/Data2.json",orient = "columns", encoding= "utf-8")

data_json2
```

### Method Data Hilang

#### `var.isna()`

Mendeteksi nilai yang hilang. Menghasilkan Dataframe Boolean.

```python
Data_Simpan.isna()
```

```python
Data_Simpan.isna().sum()
```

#### `var.notna()`

Mendeteksi nilai yang ada. Menghasilkan Dataframe Boolean.

#### `var.dropna(axis=0, how='any', subset=None)`

Digunakan untum menghapus baris / kolom yang memiliki nilai null / NaN.

```python
data_baru = Data_Simpan.dropna(axis = 0, subset=["Nama_Buah"])

data_baru
```

#### `var.fillna(value=None)`

Digunakan untuk mengisi nilai yang null / NaN

```python
data_baru = Data_Simpan.fillna(value="Unknown")

data_baru
```

```python
data_baru = Data_Simpan.fillna(value={
    "Nama_Buah" : "Unknown",
    "Diskon" : 0
})

data_baru
```

### Method Data Duplikat

#### `var.duplicated(subset=None)`

Digunakan untuk menandai baris duplikat dengan boolean

```python
data_baru = Data_Simpan.duplicated()

data_baru
```

```python
data_baru = Data_Simpan.duplicated().sum()

data_baru
```

#### `var.drop_duplicates(subset=None)`

Digunakan untuk menghilangkan data duplikat

```python
data_baru = Data_Simpan.drop_duplicates()

data_baru
```

### Method Reset Indeks & Hapus Kolom / Baris

#### `var.reset_indeks()`

```python
data_baru = data_baru.reset_index()

data_baru
```

#### `var.drop(index=None, columns=None)`

```python
data_baru = data_baru.drop(columns=["index"])
data_baru
```
